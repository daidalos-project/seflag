# Install Modules, Define funcions

```python
''' INSTALL NEEDED MODULES ETC. '''

!pip install spacy grecy
import csv
import os.path

import evaluate
import yaml
from datasets import load_dataset, DatasetDict, Dataset, concatenate_datasets
from evaluate import EvaluationModule
from tqdm import tqdm
import la_core_web_lg
import spacy
from spacy import Language
from spacy.tokens import Doc, Token
import unicodedata
import nltk
from nltk.parse import DependencyGraph, DependencyEvaluator

def annotate_greek_texts(text: str) -> list[tuple[str, str, str, str]]:
    """Adds Dependency Parsing labels to Greek texts."""
    # to install this, make sure you run this line once before:
    # !pip install spacy grecy
    nlp: Language = spacy.load("grc_proiel_trf")
    doc: Doc = nlp(text)
    tokens: list[Token] = list(doc)
    # format: base word form, relation type, target index, target word form
    return [
        (token.text, token.dep_, token.head.idx, token.head.text) for token in tokens
    ]


def annotate_latin_texts(text: str) -> list[tuple[str, str, str, str]]:
    """Adds Dependency Parsing labels to Latin texts."""
    # to install this, make sure you run these lines once before:
    # !pip install spacy
    # !pip install https://huggingface.co/latincy/la_core_web_lg/resolve/main/la_core_web_lg-any-py3-none-any.whl
    nlp: Language = la_core_web_lg.load()
    doc: Doc = nlp(text)
    tokens: list[Token] = list(doc)
    # format: base word form, relation type, target index, target word form
    return [
        (token.text, token.dep_, token.head.idx, token.head.text) for token in tokens
    ]
```

# Create latin dataset

```python
def conllu_to_text(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        conllu_data = file.read()
    sentences = conllu_data.strip().split("\n\n")
    return sentences

def conllu_to_graph(file_path):
    with open(file_path, 'r', encoding='utf-8') as file:
        conllu_data = file.read()
    sentences = conllu_data.strip().split("\n\n")
    graphs = [DependencyGraph(sentence) for sentence in sentences]
    return graphs

file_path = 'data/proiel_la_conll/caes-gal.conll'
sentences = conllu_to_text(file_path)
graphs = conllu_to_graph(file_path)
```

```python
# extracting text sentences
def extract_text(degraphs):
    texts = []
    for graph in degraphs:
        words = [node['word'] for node_id, node in graph.nodes.items() if node_id != 0]
        texts.append(" ".join(words))
    return texts

degraphs = graphs
extract_text(degraphs)
```

```python
# extracting individual words
raw_text = []
for graph in graphs:
    for node_id, node in graph.nodes.items():  # Iterate through items of graph.nodes
        if node['word'] is not None:  # Check if 'word' is not None
            raw_text.append(node['word'])  # Append the word to the list

print(raw_text)
```

```python
def convert_annot_to_conllu(annotations):
    """Convert the annotated tuples to CoNLL-U format string."""
    conllu_lines = []
    for i, (word, dep, head_idx, head_word) in enumerate(annotations, start=1):
        head = head_idx + 1  # Convert to 1-based index
        conllu_lines.append(f"{i}\t{word}\t_\t_\t_\t_\t{head}\t{dep}\t_\t_")
    return "\n".join(conllu_lines)

conll_format = convert_annot_to_conllu(annotations)
print(conll_format)
```
